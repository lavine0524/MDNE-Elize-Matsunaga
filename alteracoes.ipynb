# 🚀 Evolução do Projeto - Etapa 03 (NLP & Sentimentos)

### 1. Inteligência Artificial e Robustez do Modelo
* **Antes:** Dependência de modelos estatísticos simplificados (`pt_core_news_sm`).
* **Depois:** Implementação de uma arquitetura de carregamento resiliente. O script prioriza o modelo **`pt_core_news_lg`** (Large), garantindo maior precisão vetorial e léxica, com um mecanismo de *fallback* automático para a versão `sm` em caso de ausência do modelo pesado no ambiente.

### 2. Refinamento de Pipeline (Hybrid NLP)
* **Antes:** Confiança exclusiva no reconhecimento estatístico (NER) nativo, resultando em falsos positivos (ex: "Acho" classificado como Pessoa ou "Elize" como Empresa).
* **Depois:** Introdução de um **`Entity Ruler`** (Régua de Entidades) posicionado estrategicamente antes da camada de NER.
    * **Gabarito de Entidades:** Forçamos a classificação correta de termos sensíveis e nomes próprios (Ex: Elize Matsunaga → `PER`).
    * **Filtro de Ruído:** Implementação de uma *Blacklist* para desconsiderar verbos e expressões comuns que o modelo original confundia com entidades reais.

### 3. Engenharia de Dados e Otimização de Performance
* **Antes:** Limpeza agressiva que descartava dados úteis, removendo hashtags e menções completas (Ex: `@usuario` sumia por inteiro).
* **Depois:** Refinamento da sanitização via Regex. Agora removemos apenas os caracteres especiais (`@`, `#`), preservando o conteúdo textual (ex: `justiça_br`, `truecrime`) para enriquecer a análise de sentimentos.
* **Performance:** Upgrade para **`re.compile()`** instanciado fora do loop de execução. Isso pré-carrega as regras de limpeza na memória uma única vez, reduzindo drasticamente o tempo de processamento por post.

### 4. Análise de Sentimento e Visualização (Deep Learning)
* **Antes:** Extração bruta de entidades sem contexto emocional ou filtros de densidade léxica.
* **Depois:** * **Sentiment Analysis:** Integração com **Transformers (BERT Multilingual)** para classificação automática de polaridade em escala de 1 a 5 estrelas.
    * **Lematização:** Geração de **WordClouds** baseadas exclusivamente em *lemmas* (raízes das palavras), eliminando variações gramaticais e focando no núcleo semântico das discussões.
    * **Filtro de Exibição:** Interface de saída limpa no terminal e no CSV, ignorando automaticamente tags mapeadas como ruído ou termos irrelevantes (`IGNORAR`).
